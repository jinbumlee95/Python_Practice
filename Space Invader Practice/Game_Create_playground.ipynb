```python
import pygame
import gameObject
import spaceInvaderFunctions

pygame.init()

size = [400 , 900]
title = 'pyGame Example'
screen = pygame.display.set_mode(size)
pygame.display.set_caption(title)
clock = pygame.time.Clock()
last_time = 0




spaceInvaderFunctions.init(screen)

running = True
while running : 
    #   4-1. FPS(Frames per Sec) 설정
    clock.tick(60)
    
    #   4-2. 입력(키보드, 마우스)의 감지
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN :
            if event.key == pygame.K_x :
                spaceInvaderFunctions.shoot_raser() # x 키 누르는 액션마다 생성 (꾹 누르기 x )
                
    
    
    #   4-3. 입력, 시간에 따른 변화
    
    spaceInvaderFunctions.render_objects() # 현재 화면에 그려야 하는 오브젝트들 표기
    spaceInvaderFunctions.move_hero() # 내가 조작하는것 움직이기
    spaceInvaderFunctions.move_enemy() # 적들이 움직여야함
    spaceInvaderFunctions.move_raser() # 레이저도 움직여야함

    
    
    current_time = pygame.time.get_ticks()
    # 1초(1000 밀리초)가 지났는지 확인
    if current_time - last_time >= 1000:
        spaceInvaderFunctions.spwan_enenmy()
        ## 1초에 한번정도 호출 위치 확인해보기
        last_time = current_time


    ## 충돌 판정 구현 > raser 와 enemy 의 위치가 같다면 처리
    
    if spaceInvaderFunctions.game_end() :
        running = False
    
    #   4-4. 전사작업(그리기)
    
    #   4-5. 업데이트
# 5. 종료(x버튼을 클릭해서 종료, game over 종료)
pygame.quit()


```
